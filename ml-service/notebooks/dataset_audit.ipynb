# Dataset Quality Audit (Visualizer)

This notebook imports the `DatasetAuditor` abstraction to perform the heavy lifting and focuses purely on visualizing the results.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT))

from app.audit.audit_dataset import DatasetAuditor

sns.set_theme(style="whitegrid", palette="muted")
```

```python
df = pd.read_parquet(ROOT / "data" / "raw_feedback.parquet")
auditor = DatasetAuditor(df)
report = auditor.run_full_audit()

import json
print(json.dumps(report, indent=2))
```

## Visualizations

```python
# Plot Target Distributions if data exists
targets = ['focus_score', 'fatigue_score', 'workload_score']
if not df.empty and df['focus_score'].notnull().any():
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    for idx, target in enumerate(targets):
        sns.countplot(x=target, data=df, ax=axes[idx], palette="viridis")
        axes[idx].set_title(f"{target} Distribution")
    plt.tight_layout()
    plt.show()
```
